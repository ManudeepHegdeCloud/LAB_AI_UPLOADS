```python

# Forward Chaining in First Order Logic

# Step 1: Initial Knowledge Base
knowledge_base = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Missile", "T1"),
    ("Owns", "A", "T1")
}

# Step 2: Define inference rules
def forward_chaining(kb):

    print("=" * 55)
    print("       FORWARD CHAINING IN FIRST ORDER LOGIC")
    print("=" * 55)

    print("\nINITIAL KNOWLEDGE BASE:")
    for fact in sorted(kb):
        print("  ", fact)

    print("\nINFERENCE PROCESS:")
    step = 1
    changed = True

    while changed:
        changed = False
        new_facts = []

        # Rule 1: Missile(x) -> Weapon(x)
        for fact in list(kb):
            if fact[0] == "Missile":
                new_facts.append(
                    (("Weapon", fact[1]),
                     f"Missile({fact[1]}) -> Weapon({fact[1]})")
                )

        # Rule 2: Enemy(x, America) -> Hostile(x)
        for fact in list(kb):
            if fact[0] == "Enemy" and fact[2] == "America":
                new_facts.append(
                    (("Hostile", fact[1]),
                     f"Enemy({fact[1]}, America) -> Hostile({fact[1]})")
                )

        # Rule 3: Missile(x) AND Owns(A, x)
        #         -> Sells(Robert, x, A)
        for fact in list(kb):
            if fact[0] == "Missile":
                x = fact[1]
                if ("Owns", "A", x) in kb:
                    new_facts.append(
                        (("Sells", "Robert", x, "A"),
                         f"Missile({x}) AND Owns(A, {x}) "
                         f"-> Sells(Robert, {x}, A)")
                    )

        # Rule 4: American(p) AND Weapon(x)
        # AND Sells(p,x,r) AND Hostile(r)
        # -> Criminal(p)
        for sell in list(kb):
            if sell[0] == "Sells":
                p, x, r = sell[1], sell[2], sell[3]

                if (("American", p) in kb and
                    ("Weapon", x) in kb and
                    ("Hostile", r) in kb):

                    new_facts.append(
                        (("Criminal", p),
                         f"American({p}) AND Weapon({x}) AND "
                         f"Sells({p}, {x}, {r}) AND Hostile({r}) "
                         f"-> Criminal({p})")
                    )

        # Add only new facts
        for fact, explanation in new_facts:
            if fact not in kb:
                kb.add(fact)
                print(f"\nStep {step}:")
                print("  Rule Applied :", explanation)
                print("  New Fact     :", fact)
                step += 1
                changed = True

    return kb


# Step 3: Execute Forward Chaining
result = forward_chaining(knowledge_base)

# Step 4: Display final knowledge base
print("\n" + "=" * 55)
print("             FINAL KNOWLEDGE BASE")
print("=" * 55)

for fact in sorted(result):
    print("  ", fact)

# Step 5: Check the goal
query = ("Criminal", "Robert")

print("\n" + "=" * 55)
print("                  FINAL RESULT")
print("=" * 55)

if query in result:
    print("  GOAL PROVED SUCCESSFULLY!")
    print("  Robert is a Criminal.")
else:
    print("  GOAL COULD NOT BE PROVED.")

print("=" * 55)
```
