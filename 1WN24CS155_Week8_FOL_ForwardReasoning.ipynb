```python

# Forward Chaining in First Order Logic

# Knowledge Base (Facts)
facts = {
    ("American", "Robert"),
    ("Missile", "T1"),
    ("Owns", "A", "T1"),
    ("Enemy", "A", "America")
}

# Rules
def forward_chaining(facts):

    new_facts = True

    while new_facts:
        new_facts = False

        # Rule 1: Missiles are weapons
        for fact in list(facts):
            if fact[0] == "Missile":
                weapon = ("Weapon", fact[1])
                if weapon not in facts:
                    facts.add(weapon)
                    new_facts = True

        # Rule 2: Enemy of America is hostile
        for fact in list(facts):
            if fact[0] == "Enemy" and fact[2] == "America":
                hostile = ("Hostile", fact[1])
                if hostile not in facts:
                    facts.add(hostile)
                    new_facts = True

        # Rule 3: Robert sells all missiles owned by A
        for fact in list(facts):
            if fact[0] == "Missile":
                missile = fact[1]
                if ("Owns", "A", missile) in facts:
                    sells = ("Sells", "Robert", missile, "A")
                    if sells not in facts:
                        facts.add(sells)
                        new_facts = True

        # Rule 4: An American selling weapons to a hostile
        # nation is a criminal
        if ("American", "Robert") in facts:
            for fact in list(facts):
                if fact[0] == "Sells":
                    person, weapon, country = fact[1], fact[2], fact[3]

                    if (("Weapon", weapon) in facts and
                        ("Hostile", country) in facts):

                        criminal = ("Criminal", person)

                        if criminal not in facts:
                            facts.add(criminal)
                            new_facts = True

    return facts


# Apply Forward Chaining
result = forward_chaining(facts)

# Display all inferred facts
print("Knowledge Base after Forward Chaining:")

for fact in sorted(result):
    print(fact)

# Check the goal
if ("Criminal", "Robert") in result:
    print("\nGoal Proved: Robert is a Criminal.")
else:
    print("\nGoal Not Proved.")
```
