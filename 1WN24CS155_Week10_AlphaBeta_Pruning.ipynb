```python
def alpha_beta(node, alpha, beta, maximizing):
    if isinstance(node, int):
        print("Evaluating:", node)
        return node

    if maximizing:
        best = float('-inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, alpha, beta, False)
            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                if node[i + 1:]:
                    print("Pruned:", node[i + 1:])
                break

        return best

    else:
        best = float('inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, alpha, beta, True)
            best = min(best, value)
            beta = min(beta, best)

            if alpha >= beta:
                if node[i + 1:]:
                    print("Pruned:", node[i + 1:])
                break

        return best


# Game Tree
tree = [
    [[10, 9], [14, 18]],
    [[5, 4], [50, 3]]
]

print("===== ALPHA-BETA PRUNING =====")
print("Game Tree:", tree)
print()

result = alpha_beta(tree, float('-inf'), float('inf'), True)

print("\nOptimal Value:", result)
```
